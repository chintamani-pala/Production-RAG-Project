# Checklist before production deployement
## All checklist **should pass** before production deployement


### Security Checklist
- Input sanitization blocks prompt injection
- PII detected and masked in both input and output
- Rate limiting prevents abuse
- Pydantic validates request bodies
- Non-root Docker user
- Secrets in environment variables, never hardcoded.

### Reliability Checklist
- Model fallback chain
- Retry logic with exponential backoff
- Health check endpoint
- Graceful error responses, No stack traces to client

### Performance checklist
- response caching with TTL(time to live)
- Cache statistics endpoint
- Token budget awareness

### Observability Checklist
- langsmith tracing on every request
- Structured JSON Logging
- Metrics collection(latency, tokens, errors)
- Metrics endpoint exposed

### Deployement Checklist
- Docker container with health check
- docker-compose for local/staging
- .env.examople documented
- Tests written and passing
